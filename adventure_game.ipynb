```python
def choose_path():
    print("You stand at a crossroads.")
    print("Do you want to explore the dark forest or enter the cave?")
    choice = input("Type 'forest' or 'cave': ").lower()

    if choice == "forest":
        print("You walk into the dark forest...")
    elif choice == "cave":
        print("You step cautiously into the mysterious cave...")
    else:
        print("Invalid choice. Please try again.")
        choose_path()
```

# Task 3: Create the forest path


Actions:<br>
• Define the function forest_path() that describes the forest scenario <br>
• Provide the player with choices (follow a river or climb a tree) <br>
• Use an if-else structure to handle player choices<br>

```python
def forest_path():
    print("You walk deeper into the forest and hear the sound of running water nearby.")
    choice = input("Do you follow the river or climb a tree? (river/tree): ").lower()

    if choice == "river":
        print("You follow the river, hoping it leads you somewhere safe...")
    else:
        print("You climb the tree to scout the area from above...")
    
```

```python
forest_path()
```

# Task 4: Create the cave path

Actions: <br>
• Define the function cave_path() that describes the cave scenario<br>
• Provide the player with choices (light a torch or proceed in the dark)<br>
• Use conditionals to determine the outcome<br>

```python
def cave_path():
    print("You step into the cave, and the air grows cold and damp.")
    choice = input("Do you light a torch or proceed in the dark? (torch/dark): ").lower()

    if choice == "torch":
        print("With the torch lit, you safely navigate the cave and move closer to the treasure!")
    else:
        print("You stumble in the darkness and fall into a hidden pit. Your adventure ends here.")
```

```python
cave_path()
```

# Task 5: Run the adventure game

Actions:<br>
• Call start_game() to begin the adventure<br>
• Ensure the program runs in a loop until the player completes their journey<br>
• Provide an option to restart the game after completion<br>

```python
def start_game():
    playing = True

    while playing:
        print("\nWelcome to the Adventure!")
        name = input("What is your name, adventurer? ")
        print(f"Good luck, {name}!")

        choice = input("Do you choose the forest or the cave? (forest/cave): ").lower()

        if choice == "forest":
            forest_path()
        elif choice == "cave":
            cave_path()
        else:
            print("Invalid choice.")
            continue

        restart = input("Would you like to play again? (yes/no): ").lower()
        if restart != "yes":
            playing = False

    print("Thanks for playing!")
```

```python
start_game()
```
